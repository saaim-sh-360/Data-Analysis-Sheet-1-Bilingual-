```python
import pandas as pd
import matplotlib.pyplot as plt

#Load the dataset
df1 = pd.read_csv(r"C:\Users\Administrator\Downloads\pakistani_ecommerce_dataset.csv")

# Q1:Most popular product category
print("Q1:", df1['product_category'].value_counts().idxmax())

# Q2:Average order value
print("Q2:", round(df1['order_value'].mean(), 2))

#Q3: City with the highest number of orders
print("Q3:", df1['city'].value_counts().idxmax())

#Q4:Most used payment method
print("Q4:", df1['payment_method'].value_counts().idxmax())

#Q5:barchart showing average order value by city
df1.groupby('city')['order_value'].mean().plot(kind='bar', color='skyblue')
plt.title('Average Order Value by City')
plt.xlabel('City')
plt.ylabel('Average Order Value')
plt.xticks(rotation=0)
plt.show()
```

```python
#Load the dataset
df2 = pd.read_csv(r"C:\Users\Administrator\Downloads\student_lifestyle_dataset.csv")

# Q1:Average number of study hours
print("Q1:", round(df2['study_hours'].mean(), 2))

#Q2:Average sleep hours by gender
print("Q2:\n", df2.groupby('gender')['sleep_hours'].mean().round(2))

#Q3: Relationship between screen time and sleep hours (correlation)
print("Q3:", round(df2['screen_time'].corr(df2['sleep_hours']), 3))

#Q4:Histogram of daily transport cost
plt.hist(df2['daily_transport_cost'], bins=10, color='lightgreen', edgecolor='black')
plt.title('Distribution of Daily Transport Cost')
plt.xlabel('Daily Transport Cost')
plt.ylabel('Number of Students')
plt.show()

#Q5:Compare average internet usage by gender
print("Q5:\n", df2.groupby('gender')['internet_usage'].mean().round(2))
```
